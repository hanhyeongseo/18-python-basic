```python
# 집합 연산 사용하기
## 합집합
## 세트1 | 세트2
## set.union(세트1, 세트2) => 세트1.union(세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a | b)
print(set.union(a,b))
print(a.union(b))
```

```python
# 교집합
## 세트1 & 세트2
## set.intersection(a,b)
## a.intersection(b)

print(a & b)
print(set.intersection(a,b))
print(a.intersection(b))
```

```python
# 차집합
## a - b
## set.difference(a,b)
## a.difference(b)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a - b)
print(set.difference(a,b))
print(a.difference(b))
print("="*20)
print(b - a)
```

```python
# 대칭차집합
## 세트1 ^ 세트2
## set.symmetric_difference(a,b)
### 교집합의 반대


print(a^b)
print(set.symmetric_difference(a,b))

print(a)  #할당연산자 사용안하면 원래 a 출력
```

```python
# 집합 연산 후 할당 연산자 사용하기
## 세트1 |= 세트2
## 세트1.update(세트2)

a = {1, 2, 3, 4}
a |= {5}
print(a)

a = {1, 2, 3, 4}
a.update({5})
print(a)
```

```python
# intersection_update
## 세트1 &= 세트2
## 세트1.intersection_update(세트2)

a = {1, 2, 3, 4}
a &= {0, 1, 2, 3, 4}
print(a)

a = {1, 2, 3, 4}
a.intersection_update({0, 1, 2, 3, 4})
print(a)
```

```python
a = {1, 2, 3, 4}
a ^= {3, 4, 5, 6}
print(a)
```

```python
# 부분집합과 상위집합 확인하기

## 현재세트 <= 다른세트
## 현재셑트.issubset(다른세트)  ##is로 시작하는 함수는 True,False판별

a = {1, 2, 3, 4}
a <= {1, 2, 3, 4}
```

```python
a.issubset({1, 2, 3, 4})
```

```python
a = {1, 2, 3, 4}
a < {1, 2, 3, 4, 5}   #포함된다는 기호 수학에선 곡선형태 집합 포함 연산
```

```python
# 연습문제 26.8
a = {i for i in range(1,101) if i % 3 == 0}
b = {i for i in range(1,101) if i % 5 == 0}

print(a & b)
```

```python
# 심사문제 26.9

a, b = list(map(int,input().split()))

a = set(i for i in range(1, a + 1) if a % i == 0)
b = set(i for i in range(1, b + 1) if b % i == 0)

divisor = a & b

result = 0
if type(divisor) == set:
    result = sum(divisor)

print(result)
print(a)
print(b)
```
