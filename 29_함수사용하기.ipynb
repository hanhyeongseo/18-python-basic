```python
# return으로 함수 중간에서 빠져나오기
def not_ten(a):
    if a ==10:
        return   ## 성립시 break처럼 넘어감 
    print(a,"입니다",sep="")
```

```python
# 심사문제 29.8

x, y = map(int,input().split())

def calc(num_a, num_b):
    add = num_a + num_b
    sub = num_a - num_b
    mul = num_a * num_b
    div = num_a / num_b
    return add, sub, mul, div

a, s, m, d = calc(x, y)
print("덧셈: {0}, 뺄셈: {1}, 곱셈: {2}, 나눗셈: {3}".format(a, s, m, d))
```

```python
price, count = map(int, input("가격과 수량을 입력하세요: ").split())


def total_price(price, count):
    return price * count


result = total_price(price, count)

print("총 가격:", result, "원")
```
