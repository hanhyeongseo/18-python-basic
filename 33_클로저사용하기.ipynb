```python
# 함수 안에서 전역 변수 변경하기

x = 10
def foo():
    x = 20
    print(x)

foo()
print(x)
```

```python
# global 전역변수-----------교수님 안쓸거 같다는 말씀--------

x = 10      #x = 10선언
def foo():
    global x    #전역적으로 쓰겠다
    x = 20   # x=20으로 업데이트 됨
    print(x)

foo()
print(x)
```

```python
def foo():
    global a 
    a = 20
    print(a)

foo()
print(a)
```

```python
# 함수 안에서 함수 만들기
def print_hello():
    hello = "Hello, world!"  #지역변수 hello선언
    def print_message():   
        print(hello)
    print_message()

print_hello()
```

```python
# 지역 변수 변경하기
def A():
    x = 10  #함수안의 지역변수  /// B입장에선 전역변수
    def B():  #B라는 함수선언
        x=20   ## 지역변수 안에 지역변수

    B()  #B호출
    print(x)

A()
```

```python
# nonlocal 지역변수

def A():
    x = 10  
    def B(): 
        nonlocal x  #논로칼은 가장 가까운 해당 변수 값
        x=20 

    B() 
    print(x)

A()
```

```python
# nonlocal 이 변수를 찾는 순서
def A():
    x = 10
    y = 100   #2
    def B():
        x = 20  #4
        def C():  #6
            nonlocal x
            nonlocal y
            x = x+30
            y = y+300
            print(x)
            print(y)
        C()  #5
    B()   #3
A()  #1 / A()호출
```

```python
# global로 전역 변수 사용하기

x = 1
def A():
    x = 10
    def B():
        x = 20
        def C():
            global x  # 제일 큰 범위에 있는 변수 x를 사용 == 1
            x = x + 30
            print(x)
        C()
    B()
A()
```

```python
# 클로저 사용하기
def calc():
    a = 3  #지역변수
    b = 5
    def mul_add(x):   # 수열 만드는 방법 인듯
        return a * x + b
    return mul_add

c = calc()
print(c(1), c(2), c(3), c(4), c(5))
```

```python
# lambda로 클로저 만들기(위에 꺼랑 목적이 같음)

def calc():
    a = 3
    b = 5
    return lambda x: a * x + b

c = calc()
print(c(1))
```

```python
# 클로저의 지역변수 변경 하기 -----------공부하지 마라_----------------------
def calc():
    a = 3
    b = 5
    total = 0
    def mul_add(x):
        nonlocal total
        total = total + a * x +b
        print(total)
    return mul_add

c = calc()
c(1)
```

```python
c(2)  #값 누적 됨 8 + 11
```

```python
c(3)  # 8+11+14
```
