```python

# 문제 1. URL 분리 및 포매팅
# input() 함수를 이용하여 URL을 입력받고, 프로토콜(Protocol), 도메인(Domain), 경로(Path)로 분리하여 출력하는 프로그램을 작성하세요. (단, 입력되는 URL은 항상 프로토콜://도메인/경로 형식을 갖춘다고 가정합니다.)
# 입력값:
# https://www.example.com/login

# 출력값:
# 프로토콜: https
# 도메인: www.example.com
# 경로: login

# 요구사항:
# 1. split() 메서드를 활용하세요.
# 2. 입력함수부터 작성하세요.

url = input("URL을 입력하세요: ")

protocol, rest = url.split("://")
domain, path = rest.split("/")

print("프로토콜:", protocol)
print("도메인:", domain)
print("경로:", path)
```

```python

# 문제 2. 딕셔너리 생성 및 필터링
# 표준 입력으로 첫 번째 줄에는 회원 정보의 키(문자열), 두 번째 줄에는 값(문자열 및 숫자)이 공백으로 구분되어 입력됩니다. 이를 딕셔너리로 만들고, 키가 password인 항목을 삭제한 뒤 출력하는 프로그램을 작성하세요.
# 입력값:
# id password name age
# admin 1234 홍길동 30

# 출력값:
# {'id': 'admin', 'name': '홍길동', 'age': '30'}

# 요구사항:
# 1. dict(), zip() 함수를 사용하세요.
# 2. 반복문을 이용하세요
# 3. copy()를 이용하여 복사하여 반복문을 이용하세요.
# 4. 입력함수부터 작성하세요.

keys = input().split()
values = input().split()

member = dict(zip(keys, values))

member_copy = member.copy()

for key in member_copy:
    if key == "password":
        del member[key]

print(member)
```

```python

# 문제 3. 중첩 딕셔너리 데이터 추출
# 다음은 도서관의 책 배치 정보를 담은 딕셔너리입니다. 요구사항에 맞게 데이터를 추출하는 코드를 작성하세요.
library = { 
  "문학": [ 
    {"제목": "1984", "저자": "조지 오웰", "재고": 3}, 
    {"제목": "위대한 개츠비", "저자": "F. 스콧 피츠제럴드", "재고": 0} 
  ], 
  "과학": [ 
    {"제목": "코스모스", "저자": "칼 세이건", "재고": 5}, 
    {"제목": "이기적 유전자", "저자": "리처드 도킨스", "재고": 2} 
  ] 
}

# 요구사항:
# 1. 재고가 0인 책의 제목과 분야를 모두 출력하시오.
# 2. 전체 재고 수량의 합계를 모두 출력하세요

# 1.
for genre, books in library.items():
    for book in books:
        if book["재고"] == 0:
            print("제목:", book["제목"], "분야:", genre)


# 2.
total = 0

for books in library.values():
    for book in books:
        total += book["재고"]

print("전체 재고 수량:", total)
```

```python

# 문제 4. 조건부 숫자 치환 (FourFive)
# 입력값을 참조하여 첫 번째 정수부터 두 번째 정수까지 숫자를 출력합니다. 단, 4의 배수일 때는 Four, 5의 배수일 때는 Five, 4와 5의 공배수(20의 배수)일 때는 FourFive를 출력하고 그 외에는 숫자를 그대로 출력하세요.
# 입력값:
# 15 22

# 출력값:
# Five 16 17 18 19 FourFive 21 22

# 요구사항:
# 1. if ~ elif ~ else 문과 나머지 연산자 %를 사용하세요.
# 2. 출력값처럼 가로로 출력되게 하세요.

start, end = list(map(int, input().split()))

for i in range(start, end + 1):
    if i % 20 == 0:
        print("FourFive", end=" ")
    elif i % 4 == 0:
        print("Four", end=" ")
    elif i % 5 == 0:
        print("Five", end=" ")
    else:
        print(i, end=" ")
```

```python

# 문제 5. 마름모형 별찍기
# 아래의 출력값과 마름모 모양으로 *을 출력하는 프로그램을 작성하세요.
# 출력값: 
#    * 
#   *** 
#  *****
# *******
#  *****
#   ***
#    *  

# 요구사항:
# 중첩 for문을 사용하거나 문자열 곱셈 연산을 활용하세요.

for i in range(1, 5):
    print(" " * (4 - i) + "*" * (2 * i - 1))

for i in range(3, 0, -1):
    print(" " * (4 - i) + "*" * (2 * i - 1))
```

```python

# 문제 6. 홀수의 세제곱 리스트 만들기
# 표준 입력으로 두 개의 정수가 주어집니다. 첫 번째 수와 두 번째 수 사이의 숫자 중 홀수만 골라 세제곱한 값을 리스트로 출력하는 프로그램을 만드세요
# 입력값:
# 2 7

# 출력값:
# [27, 125, 343]

# 요구사항:
# 1. 일반 for문을 사용한 방식과 리스트 컴프리헨션을 사용한 방식 
#    2가지로 모두 작성하세요.
# 2. 입력함수부터 작성하세요.

start, end = map(int, input().split())

result = []

for i in range(start, end + 1):
    if i % 2 == 1:
        result.append(i ** 3)

print(result)
```

```python

# 문제 7. 급여 내림차순 정렬 및 포매팅
# 표준 입력으로 직원들의 급여가 쉼표(,)로 구분되어 한 줄로 입력됩니다. 입력된 급여를 높은 가격순(내림차순)으로 정렬한 뒤, 천 단위 콤마와 함께 원을 붙여 출력하세요.
# 입력값:
# 3500000,4200000,2800000,5000000

# 출력값:
# 5,000,000원
# 4,200,000원
# 3,500,000원
# 2,800,000원

# 요구사항:
# 1.sort()와 f-string을 사용하세요.
# 2. 입력함수부터 작성하세요.


salary = list(map(int, input().split(",")))

salary.sort(reverse=True)

for money in salary:
    print(f"{money:,}원")
```

```python

# 문제 8. 2차원 리스트 테두리 합 구하기
# 3x3 크기의 2차원 정수형 리스트가 표준 입력으로 주어집니다. 정중앙의 요소(인덱스 [1][1])를 제외한 나머지 8개 테두리 요소들의 합을 구하여 출력하세요.
# 입력값:
# 1 2 3 
# 4 5 6 
# 7 8 9

# 출력값:
# 40

# 요구사항:
# 1. 2차원 리스트의 인덱싱을 활용하세요.

matrix = []

for _ in range(3):
    row = list(map(int, input().split()))
    matrix.append(row)

total = 0

for i in range(3):
    for j in range(3):
        if not (i == 1 and j == 1):
            total += matrix[i][j]

print(total)
```

```python

# 문제 9. 가변 인수(*args)를 이용한 평균 이상 개수 구하기
# 여러 개의 숫자들을 입력받아 평균을 구하고, 평균보다 큰 숫자가 몇 개인지 반환하는 함수를 작성하세요.
# 입력값:
# 70 80 90 100 50

# 출력값:
# 평균: 78.0평균 이상인 수의 개수: 3

# 요구사항:
# 1.함수 이름은 analyze_scores로 지정하세요.
# 2.가변 인수 *args를 사용하세요.

def analyze_scores(*args):
    average = sum(args) / len(args)
    
    count = 0
    for score in args:
        if score > average:
            count += 1
    
    print(f"평균: {average}")
    print(f"평균 이상인 수의 개수: {count}")


scores = list(map(int, input().split()))
analyze_scores(*scores)
```

```python

# 문제 10. 키워드 인수(kwargs)를 이용한 장바구니 계산
# 상점에 등록된 상품의 가격 딕셔너리를 바탕으로 주문 금액을 계산하는 함수를 작성하세요.
# inventory = {
#  "노트북": 1500000, "마우스": 30000, "키보드": 50000, "모니터": 300000 }

# 함수를 다음과 같이 호출했을 때 결과를 출력하세요.
# checkout(노트북=1, 마우스=2)
# checkout(모니터=2, 패드=1)

# 출력값:
# 노트북 1개: 1,500,000원
# 마우스 2개: 60,000원
# 총 결제 금액: 1,560,000원

# 모니터 2개: 600,000원
# 판매하지 않는 상품입니다: 패드
# 총 결제 금액: 600,000원

inventory = {
    "노트북": 1500000,
    "마우스": 30000,
    "키보드": 50000,
    "모니터": 300000
}


def checkout(**kwargs):
    total = 0

    for product, quantity in kwargs.items():
        if product in inventory:
            price = inventory[product] * quantity
            print(f"{product} {quantity}개: {price:,}원")
            total += price
        else:
            print(f"판매하지 않는 상품입니다: {product}")

    print(f"총 결제 금액: {total:,}원")
```
