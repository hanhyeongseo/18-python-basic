```python
#회문 판별하기

word = input("단어를 입력하세요: ")

is_palindrome = True
for i in range(len(word) // 2 ):
    if word[i] != word[-1 - i]:
        is_palindrome = False
        break

print(is_palindrome)
```

```python
# N-gram  만들기

text = "Hello"

# 2-gram이므로 문자열의 끝에서 한 글자 앞까지만 반복함
for i in range(len(text) - 1):
    print(text[i], text[i + 1], sep="")
```

```python
text = "this is pythn script"
words = text.split()

for i in range(len(words) - 1):
    print(words[i], words[i + 1], sep=" ")
```

```python
text = "hello"

two_gram = zip(text, text[1:])
for i in two_gram:
    print(i[0], i[1], sep="")
```
