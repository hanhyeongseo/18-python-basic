```python
# 위치 인수를 사용하는 함수를 만들고 호출하기
def print_numbers(a, b, c):
    print(a)
    print(b)
    print(c)

print_numbers(10, 20, 30)
```

```python
# 키워드 인수와 딕셔너리 언패킹 사용하기
# 함수(**딕셔너리)
def personal_info(name, age, address):
    print("이름:", name)
    print("나이:", age)
    print("주소:", address)

x={
    "name": "홍길동",
    "age": 30,
    "address":"서울시 용산구"
}

personal_info(**x)
# ㄴ == personal_info(name="홍길동", age=30, address="서울시 용산구")   // 언팩킹
```

```python
# 키워드 인수를 사용하는 가변 인수 함수 만들기(여러개 인수 집어넣을수있다)

# def 함수이름(**매개변수): 
#     코드

def personal_info(**kwargs):    #**가 붙었기 때문에 함수에 들어오는 키워드 인수들을 딕셔너리로 묶어서 받습니다. == {"name": "홍길동"}인 딕셔너리 형태
    print(kwargs)
    print(type(kwargs))
    for kw, arg in kwargs.items():    #items가 키와 값을 둘다 가져온다.
        print(kw, ": ", arg, sep="")
        # ㄴ == print("name", ": ", "홍길동", sep="")
        
personal_info(name="홍길동")   # (name="홍길동")==키워드인수
```

```python
# 연습문제 30.6
korean, english, mathematics, science = 100,86, 81, 91

def get_max_score(*args):
    return max(args)

max_score = get_max_score(korean, english, mathematics, science)
print("높은 점수:", max_score)

max_score = get_max_score(english, science)
print("높은 점수:", max_score)
```

```python
# 심사문제 30.7  (76 82 89 84)
korean, english, mathematics, science = map(int,input().split())

def get_min_max_score(*scores):                 #튜플
    return min(scores), max(scores)

def get_average(**scores):                      #딕셔너리
    return sum(scores.values()) / len(scores)

min_score,max_score = get_min_max_score(korean, english, mathematics, science)
average_score = get_average(korean=korean, english=english, mathematics=mathematics, science=science)
print("낮은 점수: {0:.2f}, 높은 점수:{1:.2f}, 평균 점수:{2:.2f}".format(min_score, max_score, average_score))

min_score, max_score = get_min_max_score(english,science)
average_score = get_average(english=english, science=science)
print("낮은 점수: {0:.2f}, 높은 점수:{1:.2f}, 평균 점수:{2:.2f}".format(min_score, max_score, average_score))
```

```python
# 메뉴 가격
menu = {
    '아메리카노': 3000,
    '라떼': 4000,
    '스무디': 4500,
    '에이드': 5000
}

# 메뉴별 원가
cost = {
    '아메리카노': 1000,
    '라떼': 1500,
    '스무디': 2000,
    '에이드': 2000
}

# 하루 알바비
alba = 100000

# 일별 매출과 원가
daily_data = {}
daily_cost = {}

# 월별 매출과 원가
monthly_data = {}
monthly_cost = {}

# 영업한 날짜 저장
working_days = {}


# 주문 함수
def order(date, item, cnt):
    sales = menu[item] * cnt
    item_cost = cost[item] * cnt

    month = date[:7]

    # 일별 매출과 원가 누적
    daily_data[date] = daily_data.get(date, 0) + sales           # 해당 날짜 매출이있으면 가져오고 없으면 0  // 누적
    daily_cost[date] = daily_cost.get(date, 0) + item_cost

    # 월별 매출과 원가 누적
    monthly_data[month] = monthly_data.get(month, 0) + sales
    monthly_cost[month] = monthly_cost.get(month, 0) + item_cost

    # 영업일 기록
    working_days[date] = 1

    return sales


# 일별 매출과 수익
def daily_report(date):
    sales = daily_data.get(date, 0)
    item_cost = daily_cost.get(date, 0)

    profit = sales - item_cost - alba

    return sales, item_cost, profit


# 월별 매출과 수익
def monthly_report(month):
    sales = monthly_data.get(month, 0)
    item_cost = monthly_cost.get(month, 0)

    # 해당 월의 영업일 수
    days = 0

    for date in working_days:
        if date[:7] == month:
            days += 1

    total_alba = alba * days

    profit = sales - item_cost - total_alba

    return sales, item_cost, total_alba, profit


# 주문할 날짜
date = input("주문 날짜를 입력하세요 (예: 2026-09-21): ")

# 하루에 판매할 음료 종류의 개수
drink_count = int(input("판매할 음료 종류의 개수: "))

# 여러 종류의 음료 입력
for i in range(drink_count):

    item = input("음료 이름을 입력하세요: ")

    cnt, = map(int, input("음료 수량을 입력하세요: ").split())

    payment = order(date, item, cnt)

    print(item, cnt, "잔")
    print("결제 금액:", payment, "원")


# 다른 날짜의 주문도 입력할지 확인
more = input("\n다른 날짜의 주문도 입력하시겠습니까? (y/n): ")

while more == "y":

    date = input("주문 날짜를 입력하세요 (예: 2026-09-21): ")

    drink_count = int(input("판매할 음료 종류의 개수: "))

    for i in range(drink_count):

        item = input("음료 이름을 입력하세요: ")

        cnt, = map(int, input("음료 수량을 입력하세요: ").split())

        payment = order(date, item, cnt)

        print(item, cnt, "잔")
        print("결제 금액:", payment, "원")

    more = input("\n다른 날짜의 주문도 입력하시겠습니까? (y/n): ")


# 일별 전체 출력
print("\n===== 일별 매출 현황 =====")

for date in daily_data:

    sales, item_cost, profit = daily_report(date)

    print(date)
    print("매출:", sales, "원")
    print("원가:", item_cost, "원")
    print("알바비:", alba, "원")
    print("수익:", profit, "원")
    print()


# 월별 전체 출력
print("===== 월별 매출 현황 =====")

for month in monthly_data:

    sales, item_cost, total_alba, profit = monthly_report(month)

    print(month)
    print("매출:", sales, "원")
    print("원가:", item_cost, "원")
    print("알바비:", total_alba, "원")
    print("수익:", profit, "원")
    print()
```
