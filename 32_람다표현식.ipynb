```python
# filter
# filter(함수, 반복가능한객체)

def f(x):
    return x > 5 and x < 10

a = [8, 3, 2, 10, 15, 7, 1, 9, 0, 11]

# f(8)
list(filter(f, a))  # 함수명 f 만 적어야함 f(x)라고 적으면 안됨!!
```

```python
#reduce==(피보나치 수열과 유사)

# import random
# random.randint(1, 10)

# from random import randint
# randint(1,10)

from functools import reduce #랜덤 가져오듯이 많은 함수 들어있는 모듈 을 들고 와 그안에있는 함수 사용할때 리듀쓰//
# reduce(함수, 반복가능한 객체)

def f(x, y):
    return x + y

a = [1, 2, 3, 4, 5]
reduce(f, a)  # 1+2=3, 3+3=6, 6+4=10....=== ((((1+2)+3)+4)+5)

```

```python
a = [1, 2, 3, 4, 5]
from functools import reduce  #functools라이브러리에 들어있는 함수인 reduce들고옴  / from 다음 모듈이 나오고 import 뒤에 변수가 나옴

reduce(lambda x, y: x + y, a)
```

```python
# 연습문제 32.4
files = ["font", "1.png", "10.jpg", "11.gif", "2.jpg", "3.png", "table.xslx", "spec.docx"]
print(list(filter(lambda x: ".jpg" in x or ".png" in x, files)))
```

```python
# 연습문제 32.4
files = ["font", "1.png", "10.jpg", "11.gif", "2.jpg", "3.png", "table.xslx", "spec.docx"]
print(list(filter(lambda x: x.find(".jpg") != -1 or x.find(".png") != -1, files)))
# .jpg나 .png를 찾았다 → 0 이상의 숫자 ====(10.jpg의 경우 .jpg 를 find 할때 인덱스가 2로 나옴)
# 못 찾았다 → -1
# 따라서 .jpg나 .png가 없지 않다 (2번 부정문 == 존재하냐?) 를 묻는 의미
```

```python
# 심사문제 32.5  (1.jpg 10.png 11.png 2.jpg 3.png)
files = input().split()
files = list(map(lambda x: "{:0>3}.{}".format(x.split('.')[0], x.split('.')[1]),files))
print(*files)
```

```python
# 심사문제 32.5  (1.jpg 10.png 11.png 2.jpg 3.png)
files = input().split()

files = list(map(lambda x: x.split('.')[0].zfill(3) + '.' + x.split('.')[1], files))

print(*files)
```
