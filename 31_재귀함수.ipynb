```python
# 재귀호출에 종료 조건 만들기

def hello(count):
    if count == 0:
        return  # 0이면 리턴

    print("hello world", count)  #(count)기본설정=줄바꿈

    count -= 1
    hello(count)
```

```python
hello(5)
```

```python
# 재귀호출로 팩토리얼 구하기
def factorial(n):
    if n ==1:
        return 1   #1이면 1돌려주고 리턴
    return n * factorial(n - 1) # 내여쓰기를 함으로써 마치 else 생략한것처럼 해석

print(factorial(5))
```

```python
# 5 : 5 * factorial(5-1)  ##=========리턴 했지만 그 안에 factorial이 아직 있음 다시 factorial 함수 풀어줘야함
# 4 : 5 * 4 * factorial(4-1)
# 3 : 5 * 4 * 3 * factorial(3-1)
# 2 : 5 * 4 * 3 * 2 * factorial(2-1)
# 1 : 5 * 4 * 3 * 2 * 1   ########### 1 == return 1


print(factorial(5))
```

```python
# 연습문제 31.4
def is_palindrome(word):
    if len(word) < 2:
        return True
    if word[0] != word[-1]:
        return False
    return is_palindrome(word[1:-1])   # word[1:-2+1]

print(is_palindrome("hello"))
print(is_palindrome("level"))

```

```python
# 심사문제 31.5(피보나치 수열)

def fib(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fib(n - 1) + fib(n - 2)  # 반복문 처럼 작동


n = int(input())
print(fib(n))
```
