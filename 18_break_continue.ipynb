```python
# break로 반복문 끝내기

i = 0
while True:
    print(i)
    i += 1
    if i == 10:
        break
```

```python
# for에서 break로 반복문 끝내기
for i in range(10000):
    print(i)
    if i == 10:
        break
```

```python
# continue로 코드 실행 건너 뛰기

for i in range(20):
    if i % 2 == 0:
        continue # 다시 반복문 시작으로 다시 복귀 (for 직행 열차)
    print(i)
```

```python
# while 반복문에서 continue로 코드 실행 건너 뛰기
i = 0
while i < 20:
    i += 1
    if i % 2 == 0:  # 짝수는 여기 if 에 걸려서 continue 만나고 다시 첨으로 ㄱㄱ
        continue
    print(i)
    
```

```python
# 입력한 횟수대로 반복하기
count = int(input())

i = 0
while True:
    print(i)
    i += 1
    if i == count:
        break
```

```python
count = int(input())

for i in range(count + 1):
    if i % 2 == 0:
        continue
    print(i)
```

```python
# 연습문제 18.5

i = 0
while True:
    if i % 10 != 3:
        i += 1
        continue
    if i > 73:
        break
    print(i, end=" ")
    i += 1
```

```python
# 심사문제 18.6
start, stop = map(int, input().split())

i = start

while True:
    if i % 10 == 3:
        i += 1
        continue

    if i > stop:
        break

    print(i, end=" ") ########################
    i += 1
```
