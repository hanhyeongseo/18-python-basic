```python
# 조건문에 숫자 지정하기
if 0:
    print("참")
else:
    print("거짓")
#-------------------------------
if 1:
    print("참")
else:
    print("거짓")
#-------------------------------
if 0x1ff:
    print("참")
else:
    print("거짓")
#-------------------------------
if 0b1000:
    print("참")
else:
    print("거짓")
#---------------------------
if 13.5:
    print("참")
else:
    print("거짓")
```

```python
# if조건문에 문자열 지정하기
if "hello":
    print("참")
else:
    print("거짓")

if "":                          ###빈문자열은 거짓
    print("참")
else:
    print("거짓")

if "  ":                       ####공백은 문장열로 인식후 참
    print("참")
else:
    print("거짓")
```

```python
# 조건식을 여러개 지정하기
x = 10
y = 20

if (x == 10) and (y == 20):      ####판다스에는 이렇게 괄호 필수
    print("참")
else:
    print("거짓")
```

```python
# 심사문제 14.7  (89 72 93 82)
scores = list(map(int,input().split()))
average = sum(scores) / 4

if any(score > 100 for score in scores):
    print("잘못된 점수입니다.")

if average >= 80:
    print("합격")
else:
    print("불합격")
```

```python
# 심사문제 14.7  (99 85 101 90) 교수님 버전
a,b,c,d = list(map(int, input().split()))

if 0 <= a <= 100 and 0 <= b <= 100 and 0 <= c <= 100 and 0 <= d <= 100:
    avg = (a + b + c + d) / 4
    if avg >= 80:
        print("합격")
    else:
        print("불합격")
else:
    print("잘못된 점수")

```
