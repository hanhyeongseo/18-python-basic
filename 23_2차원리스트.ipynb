```python
# 심사문제 23.7
row, col = map(int, input().split())   #몇 행 몇 열 인지 인풋

matrix = []                          # matrix는 리스트임을 선언

for i in range(row):               #행// matrix라는 리스트에 i번째 열에 리스트 형식의 인풋값을 append해줌, 을 반복
    matrix.append(list(input()))

for i in range(row):                #i 번쨰 열 설명시작(리스트 형식이므로 하나의 열 만 존재하는 상태): 
    for j in range(col):                 #j 열에서 

        if matrix[i][j] == "*":
            continue

        count = 0

        for x in range(i - 1, i + 2):                #범위 안에 지뢰 갯수 세기
            for y in range(j - 1, j + 2):

                if 0 <= x < row and 0 <= y < col:
                    if matrix[x][y] == "*":
                        count += 1

        matrix[i][j] = str(count)

for i in range(row):
    print("".join(matrix[i]))
```

```python
# 심사문제 23.7

row, col = map(int, input().split())
# row = 행의 개수, col = 열의 개수

matrix = []
# matrix라는 빈 리스트를 만듦

for i in range(row):
    # row의 개수만큼 반복 → 행의 개수만큼 반복
    # i = 현재 몇 번째 행인지 나타냄

    matrix.append(list(input()))
    # 한 줄을 input()으로 입력받음
    # list()를 이용해서 입력받은 문자열을 문자 하나씩 나누어 리스트로 만듦

for i in range(row):
    # 모든 행을 하나씩 검사
    # i = 현재 검사하고 있는 행의 번호

    for j in range(col):
        # 현재 행에서 모든 열을 하나씩 검사
        # matrix[i][j]

        if matrix[i][j] == "*":
            # 현재 위치가 지뢰(*)라면

            continue
            # 아래의 지뢰 개수 계산을 하지 않고 다음 칸으로 넘어감
            # 지뢰 자체를 숫자로 바꾸면 안 되기 때문

        count = 0
        # 현재 칸 주변에 있는 지뢰의 개수를 저장할 변수, 시작 = 0

        for x in range(i - 1, (i + 1) +1):
            # 현재 행(i)을 기준으로 위, 현재 행, 아래 행을 검사

            for y in range(j - 1, j + 2):
                # 현재 열(j)을 기준으로 왼쪽, 현재, 오른쪽 열을 검사
                #
                # 따라서 x와 y를 같이 사용하면 현재 칸을 중심으로 3 × 3 범위를 검사하게 됨

                if 0 <= x < row and 0 <= y < col:
                    # x와 y가 matrix의 범위를 벗어나지 않는지 확인
                    #
                    # 가장자리 칸에서는 i-1이나 j-1 때문에
                    # 범위를 벗어날 수 있기 때문에 확인하는 것
                    #
                    # x는 0 이상 row 미만이어야 함
                    # y는 0 이상 col 미만이어야 함

                    if matrix[x][y] == "*":
                        count += 1

        matrix[i][j] = str(count)
        # 현재 칸에 주변 지뢰의 개수를 저장
        #
        # count는 숫자(int)이기 때문에
        # str(count)를 사용해서 문자열로 변환
        # matrix에는 문자 형태로 저장해야
        # 나중에 "*"와 함께 출력하기 편함

for i in range(row):
    # 모든 행을 하나씩 출력

    print("".join(matrix[i]))
    # matrix[i]는 한 행을 이루는 리스트
    #
    # 예:
    # ['1', '2', '*', '1']
    #
    # "".join()을 사용하면
    # ['1', '2', '*', '1']
    #       ↓
    # "12*1"
    #
    # 즉, 리스트의 요소들을 공백 없이 하나의 문자열로 합쳐서 출력
```

```python
## 23_2차원리스트 심사문제

col, row = map(int,(input()).split())

# 방향 (상, 하, 좌, 우, 좌상, 우상, 좌하, 우하)///몇행 몇열인지를 인덱스 번호를 부가하여 사용하기도 함
dx = [-1, 1, 0, 0, -1, -1, 1, 1]  
dy = [0, 0, -1, 1, -1, 1, -1, 1]  

matrix = []
for i in range(row):
    matrix.append(list(input()))

for i in range(row):
    for j in range(col):
        if matrix[i][j] == "*":
            print("*", end = "")
        else:
            count = 0
            for k in range(8): # 방향 안쓰고하면 일일히 전부 체크해야됨
                nx = i + dx[k]
                ny = j + dy[k]    
                if 0 <= nx < row and 0 <= ny < col:
                    if matrix[nx][ny] == "*":
                        count = count + 1

            print(count, end = "")
    print()

```
