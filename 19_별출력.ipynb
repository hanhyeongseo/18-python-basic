```python
# 중첩 루프 사용하기  (중첩 for 문)
for i in range(5):
    for j in range(5):
        print("j:", j, sep="", end=" ")
    print("i:", i, "\\n", sep="")
```

```python
for i in range(1,11):
    if i % 3 == 0:
        continue
    print(i)
```

```python
# 사각형으로 별 출력하기
for i in range(5):                    #i랑 j는 상관없는 변수여서 _ 이런식으로 표현해도 노상관
    for j in range(5):
        print("*", end="")
    print()   #print안에 아무것도 안적으면 end 파라미터의 디폴트값이 \n이기 때문임
```

```python
# 사각형 모양 바꾸기
for i in range(3):
    for j in range(7):
        print("*", end="")
    print()  

```

```python
# 계단식으로 별 출력하기
for i in range(5):
    for j in range(5):
        if j <= i:
            print("*", end="")
    print()
```

```python
# 대각선으로 별 출력하기
for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end="")
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end="")
        else:
            print("  ", end="")
    print()
```

```python
# 연습문제 19.5
for i in range(5):
    for j in range(5):
        if j < i:
            print(" ", end="")
        else:
            print("*", end="")
    print()
```

```python
#심사문제 19.6
a = int(input())

for i in range(a):
    for j in range(2 * a - 1):
        if j < a - i - 1 or j > a + i - 1:
            print(" ", end="")
        else:
            print("*", end="")
    print()
```

```python
#심사문제 19.6
n = int(input())

for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
#심사문제 19.6
num = int(input())

for i in range(num):
    print(" " * (num-1-i) + "*" * (2 * i + 1))
```
