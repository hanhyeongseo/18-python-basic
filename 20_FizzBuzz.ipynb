```python
# 1부터 100까지 숫자 출력하기
for i in range(1, 10+1):
    print(i)
```

```python
# 3의 배수일때 Fizz, 5의 배수일때 Buzz
for i in range(1, 20+1):
    if i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
for i in range(1, 20+1):
    if i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    elif i % 3 == 0 and i % 5 == 0:   # i % 15 ==0
        print("FizzBuzz")
    else:
        print(i)
```

```python
for i in range(1, 20+1):
    if i % 3 == 0 and i % 5 == 0:   # i % 15 ==0
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 코드 단축하기
for i in range(1, 30+1):
    print("Fizz" * (i % 3 == 0) + "Buzz" * (i % 5 == 0) or i) ## or 앞이 구라면  i 출력

    # i%3==0이 성립(True)되면 1이되어 Fizz*1 이 된다.
```

```python
# 연습문제 20.7
for i in range(1, 101):
    if i % 2 == 0 and i % 11 == 0:   
        print("FizzBuzz")
    elif i % 2 == 0:
        print("Fizz")
    elif i % 11 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 심사문제 20.8
small, big = map(int, input().split())

for i in range(small, big+1):
    if i % 5 == 0 and i % 7 == 0:   
        print("FizzBuzz")
    elif i % 5 == 0:
        print("Fizz")
    elif i % 7 == 0:
        print("Buzz")
    else:
        print(i)

#가로 출력 원하면 프린트 마다 end=" "
```

```python
# 교수님 제작 문제
# 중첩 for 이용
test = [
    [1, 3, 5, 7, 9],
    ["a", "b", "c", "d", "e"],
    ["1", "2", "3", "4", "5"]
]

for i in test:
    for j in i:
        print(j, end=" ")
```
