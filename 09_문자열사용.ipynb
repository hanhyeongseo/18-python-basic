```python
# 여러 줄로 된 문자열 사용하기
hello = '''Hello, world
안녕하세요
Python입니다'''
print(hello)

# 여기서 hello == 객체라고도 표현함
# 작은 따옴표나 큰 따옴표 3개여야지 위의 형식 사용가능
```

```python
# 문자열 안에 작은따옴표나 큰따옴표 포함하기
s = "Python isn's difficult"
print(s)
```

```python
s = 'He said "Python is easy"'
print(s)
```
