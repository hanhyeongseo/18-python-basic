```python
# # while 반복문 사용하기

# 초기식 
# while 조건식:
#     반복할코드
#     변화식

i = 0
while i < 10:
    print("hello world")
    i += 1
```

```python
# 초깃값을 1부터 시작하기
i = 1
while i <= 5:
    print("hello world", i)
    i += 1
```

```python
# 초깃값을 감소시키기

i = 5
while i > 0:
    print("hello", i)
    i -= 1
```

```python
count = int(input())

i = 0
while i < count:
    print("hello", i)
    i += 1
```

```python
# 심사문제 17.6
money = int(input())
count = 1

while money >= 1350:
    money -= 1350
    print(money)
    count += 1 # 이줄은 의미가없음 = while에서 계속 빼기를 반복하기 때문
```

```python
# 심사문제 17.6
bal = int(input())
fare = 1350

while bal >= fare:
    bal -= fare
    print(bal)
```
