```python
# 반복문에서 변수의 변화 확인
for i in range(10):
    print("hello world", i) # i는 임의변수
```

```python
for number in range(5):
    print("hello world", number)  ##시퀀스객체나 이터러블객체에서 꺼내쓰는...반복문 만들때 적극활용
```

```python
# 시작하는 숫자와 끝나는 숫자 지정하기
for a in range(5, 12):
    print(a)
```

```python
# 증가폭 사용하기
for b in range(0, 10, 2):
    print(b)
```

```python
# 숫자를 감소시키기
for c in range(10, 0, -1): #꺼내쓰고 다음으로 넘어가는객체를 이터러블객체라고 함  # -1 안적으면 기본값이 +1이기때문에 아무값도 나오지않는에러발생
    print(c) 
```

```python
for i in reversed(range(10)):
    print(i)#########################----생각은 10부터일것같았지만 완전히 래인지의 리버스형태
```

```python
# 입력한 횟수대로 반복하기
count = int(input())

for i in range(count):
    print('hello', i)
```

```python
# 시퀀스 객체로 바꾸기

a = [10, 20, 30, 40, 50]  ##문자열 시퀀스 리스트 등등 넣을수있음
for i in a:
    print(i)
```

```python
fruits = ["apple", "orange", "grape"]

for fruit in fruits:
    print(fruit)
```

```python
for letter in "python":
    print(letter, end = " ")  #end기본값 == 줄바꿈
```

```python
for letter in reversed("python"):
    print(letter, end="")
```

```python
#연습문제 16.5
x = [49, -17, 25, 102, 8, 62, 21]
for i in x:
    print(i * 10, end=" ")
```
