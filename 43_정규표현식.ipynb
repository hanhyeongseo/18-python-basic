```python
# 문자열 판단하기
import re

re.match("Hello", "Hello, world!")
re.match("Python", "Hello, world!")
```

```python
# 문자열이 맨 앞에 오는지 맨 뒤에 오는 지 판단하기
# ^문자열 === 이 문자열로 시작할때
# 문자열$ === 이 문자열로 끝날때

re.search("^Hello", "Hello world!")
re.search("world$", "Hello world!")
```

```python
# 지정된 문자열이 하나라도 포함되는지 판단하기
#매치는 처음부터고 서치는 중간부터 사용가능

re.match("hello|world", "hello")
```

```python
# # 범위판단하기
# [0-9]*
# [0-9]+ 

print(re.match("[0-9]*", "1234"))  #전부다
print(re.match("[0-9]+", "1234"))  #1개이상 반드시 있어야함
print(re.match("[0-9]+", "abcd"))  
```

```python
print(re.match("a*b", "b")) #0개 이상있는지 ==없어도 됨
print(re.match("a+b", "b"))  #a가 한개이상은 있어야함
print(re.match("a*b", "aab"))
print(re.match("a+b", "aab"))
```

```python
# 문자가 한 개만 있는지 판단하기 (0개 또는 1개 있는지 판별) ######################자주씀
# 문자?
# [0-9]?

print(re.match("abc?d","abd"))
print(re.match("abc?d","abcd"))
print(re.match("abc?d","abccd"))
print(re.match("ab[0-9]?c", "ab3c"))
# print(re.match("ab[0-9]?c", "abc"))
print(re.match("ab.d", "abxd"))  # .은 한개는 있어야한다.

```

```python
# 문자개수 판단하기+++++++++++++++++++++++++++++발목잡힌다 잘해라+++++++++++++
# 문자{개수}
# (문자열){개수}

print(re.match("h{3}","hhhello"))
print(re.match("h{3}","hhhhello"))
print(re.match("hello{3}","hellohellohelloworld"))  # ooo가 나오는 지 확인하는거
print(re.match("(hello){3}","hellohellohelloworld")) # (hellohellohello)가 나오는지 확인
```

```python
print(re.match("[0-9]{3}-[0-9]{4}-[0-9]{4}", "010-1234-1234"))
print(re.match("[0-9]{3}-[0-9]{4}-[0-9]{4}", "010-1234-123"))
```

```python
# (문자){시작개수, 끝개수}
# (문자열){시작개수, 끝개수}
# [0-9]{시작개수, 끝개수}

print(re.match("[0-9]{2,3}-[0-9]{3,4}-[0-9]{4}", "02-123-1234"))
print(re.match("[0-9]{2,3}-[0-9]{3,4}-[0-9]{4}", "010-1234-1234"))
print(re.match("[0-9]{2,3}-[0-9]{3,4}-[0-9]{4}", "02-12-1234"))
```

```python
# 숫자와 영문 문자를 조합해서 판단하기
# a-z
# A-Z

#[a-zA-Z0-9]==한칸 의미

print(re.match("[a-zA-Z0-9]+","eHello1234"))
print(re.match("[A-Z0-9]+","hello"))
```

```python
# 가-힣(우리나라 글자 법위)
re.match("[가-힣]+", "홍길동")
```

```python
# 특정 문자 범위에 포함되지 않는지 판단하기
# [^범위]
# [^범위]+ 

print(re.match("[^A-Z]+", "Hello"))
print(re.match("[^A-Z]+", "hello"))
```

```python
# ^[범위]
# ^[범위]+   
# 시작할때

print(re.search("^[A-Z]+", "Hello"))
```
