```python
# 값을 여러개 출력하기
print(1, 2, 3)
print("hello", "python")
```

```python
##sep 로 값 사이에 문자 넣기
print(1, 2, 3)
print(1, 2, 3, sep=", ")
print(1, 2, 3, sep=" ")##===============기본값
print(1, 2, 3, sep="")
```

```python
print("hello", "python")
print("hello", "python", sep="")
```

```python
print(1920, 1080, sep="x")
```

```python
#줄바꿈 활용
print(1, 2, 3)
print(1, 2, 3, sep="\n") 
```

```python
print("1\n2\n3")
```

```python
## end 사용하기
print(1)
print(2)
print(3)
```

```python
print(1,end="")
print(2,end="")
print(3,end="")
#end= 줄바꿈이 기본값이다. ==== \n

#따라서 파라미터도 수정이가능하다!!
```

```python
##연습문제 7.4
year = 2000
month = 10
day = 27
hour = 11
minute = 43
second = 59

print(year, month, day, sep='/', end=" ")
print(hour, minute, second, sep=":")
```

```python
##연습문제 7.5             (입력: 1999 12 31 10 37 21)
year, month, day, hour, minute, second = input().split()
print(year, month, day, sep="-", end="T")
print(hour, minute, second, sep=":")
```
