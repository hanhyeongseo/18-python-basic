```python
# 리스트 안에 리스트 추가하기
a = [10, 20, 30]
a.append([500, 600])
print(a)
len(a)
```

```python
# 리스트 확장하기
a = [10, 20, 30]
a.extend([500, 600])  #리스트형식만 extend가능
print(a)
```

```python
len(a)
```

```python
# 리스트의 특정 인덱스에 요소 추가하기
a = [10, 20, 30]
a.insert(2, 500)  #2번 인덱스 자리에 추가
print(a)
```

```python
a = [10, 20, 30]
a.insert(0, 500)    # 쉘에서는 .이 속성을 가져올때 사용/// 터미널에서는 현재 파일
print(a)
```

```python
a = [10, 20, 30]
a.insert(len(a), 500)
a
```

```python
a = [10, 20, 30]
a.insert(1, [500, 600])
a
```

```python
a = [10, 20, 30]
a[1:1] = [500, 600]
a
```

```python
a = [10, 20, 30]
a[1:2] = [500, 600]
a
```

```python
# 리스트에서 특정 인덱스의 요소를 삭제하기 $$$
a = [10, 20, 30]
print(a.pop())    ## 자주씀   #뺸 값을 출력할때 사용하는 방식!!!!!!!!!!!!!!!!
```
