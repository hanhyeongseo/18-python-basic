```python
#인덱스와 요소를 함께 출력하기 - enumerate() ##인덱스와 요소를 짝지어서 보여줌 >> 언팩킹 ㄱㄱ
a = [38, 21, 53, 61, 19]

for index, value in enumerate(a):
    print(index, value)
```

```python
#인덱스와 요소를 함께 출력하기 - enumerate() ##인덱스와 요소를 짝지어서 보여줌 >> 언팩킹 ㄱㄱ+++++++++++++++++자주씀
a = [38, 21, 53, 61, 19]

for index, value in enumerate(a):
    print(index + 1, value)
```

```python
# # 리스트에서 map사용하기
# list(map(함수, 리스트))
# tuple(map(함수, 리스트))

a = [1.2, 2.5, 3.7, 4.6]
for i in range(len(a)):     #i = 0 1 2 3
    a[i] = int(a[i])       # = 오른쪽부터 해석

print(a)
```

```python
# 심사문제 22.10
small, big = map(int,input().split())

a = [2**i for i in range(small, big + 1)]
del a[1]
del a[-2]
print(a)
```

```python
# 심사문제 22.10 = 교수님 버전

small, big = map(int,input().split())

result = [2**val for i,val in enumerate(range(small, big + 1))if i != 1 and i != big - small -1]
print(result)
```
