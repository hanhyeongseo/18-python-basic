```python
# 반복문으로 리스트의 ㅁ요소를 모두 출력하기================자주씀===============================

a = [38, 21, 53, 61, 19]
for i in a:   #a = 5줄
    print(i)
```

```python
#인덱스와 요소를 함께 출력하기 - enumerate() ##인덱스와 요소를 짝지어서 보여줌 >> 언팩킹 ㄱㄱ
a = [38, 21, 53, 61, 19]

for index, value in enumerate(a):
    print(index, value)
```

```python
#인덱스와 요소를 함께 출력하기 - enumerate() ##인덱스와 요소를 짝지어서 보여줌 >> 언팩킹 ㄱㄱ+++++++++++++++++자주씀
a = [38, 21, 53, 61, 19]

for index, value in enumerate(a):
    print(index + 1, value)
```

```python
a = [38, 21, 53, 61, 19]

for index, value in enumerate(a, start =1):
    print(index, value)
```

```python
# while 반복문으로 요소 출력하기

a = [38, 21, 53, 61, 19]
i = 0
while i < len(a):
    print(a[i])
    i += 1
```

```python
a = [38, 21, 53, 61, 19]
i = 0
while i <= len(a):      # len(5)가 없음
    print(a[i])
    i += 1
```

```python
# 리스트의 가장 작은 수, 가장 큰 수 구하기
a = [38, 21, 53, 61, 19]

# 데이터 베이스의 집계함수 == 여러데이터를 하나로 만드는 함수// +평균, 표준편차 등등
print(max(a))
print(min(a))
print(sum(a))

```

```python
a = [38, 21, 53, 61, 19]

smallest = a[0]
for i in a:
    if i < smallest:
        smallest = i

print(smallest)
```

```python
a = [38, 21, 53, 61, 19]

# a 에서 첫번째꺼 들고와서 for문 싸이클 돌리고 다음 2번쨰꺼 들고와서 for문 싸이클돌리는 형식 반복

largest = a[0]
for i in a:
    if i > largest:
        largest= i

print(largest)
```
