```python
# 딕셔너리 조작하기
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}
x.setdefault("e")
print(x)
```

```python
x.setdefault("f", 100)
print(x)
```

```python
# 딕셔너리에서 키의 값 수정하기
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}
x.update(a=90)
print(x)
```

```python
x.update(e=50)
print(x)
```

```python
x.update(a=900, f=60) ##########################################
print(x)
```

```python
y = {
    1: "one",
    2: "two"
}

y.update({
    1: "ONE",
    2: "TWO"
})

print(y)
```

```python
y.update(
    [
        [2,"TWO"],
        [4,"FOUR"]
    ]
)
print(y)  #2차원배열
```

```python
y.update(
    zip([1, 2], ["one", "two"])
)

print(y)
```

```python
# 딕셔너리에서 키-값 쌍 삭제하기
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}

x.pop("a")
print(x)
```

```python
x.pop("z", 0)  #에러 처리를 pop을 이용하여 바로 할수있음(z키없어서 실행시 에러 생김)
```

```python
del x["b"]
print(x)
```

```python
# 딕셔너리에서 임의의 키-값 쌍 삭제하기
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}

print(x.popitem())   #마지막요소 제거
print(x)
```

```python
# 딕셔너리에서 모든 키-값 쌍을 삭제하기
print(x)
x.clear()

print(x)
```

```python
# 딕셔너리에서 키의 값을 가져오기 (==indexing)
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}
print(x["a"])
# print(x["z"])

print(x.get("a"))
print(x.get("z", 0))  #없으면 None대신 0을 반환

```

```python
# 딕셔너리에서 키-값 쌍을 모두 가져오기
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}
type(x.items())
print(x.items())
print(list(x.items()))
```

```python
#keys()--a모든 키를 가져옵니다.

x.keys()
```

```python
# values() => 모든 값을 가져옵니다.
x.values()
list(x.values())
```

```python
# 리스트와 튜플로 딕셔너리 만들기
keys = ["a", "b", "c", "d"]
x = dict.fromkeys(keys) ################사용많음
x
```
