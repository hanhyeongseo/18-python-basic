```python
y = {
    1: "one",
    2: "two"
}

y.update({
    1: "ONE",
    2: "TWO"
})

print(y)
```

```python
y.update(
    [
        [2,"TWO"],
        [4,"FOUR"]
    ]
)
print(y)  #2차원배열
```

```python
# 딕셔너리 표현식 사용하기 (== 리스트 컨프리헨션)
# {키:값 for 키, 값 in 딕셔너리}

keys = ["a", "b", "c", "d"]
imsi = dict.fromkeys(keys)
imsi

x = {key:value for key, value in imsi.items()}
x
```

```python
# 딕셔너리 표현식에서 if 조건문 사용하기
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}

for key, value in x.items():
    if value == 20:
        del x[key]
print(x)
```

```python
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}

del x["b"]
print(x)
```

```python
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}
y= x.copy()

for key, value in y.items():
    if value == 20:
        del x[key]

print(x)
```

```python
x = {
    "a": 10,
    "b": 20,
    "c": 30,
    "d": 40
}

x = {key: value for key,  value in x.items() if value != 20} #필터링
```

```python
# 심사문제 25.8
# alpha bravo charlie delta
# 10 20 30 40

keys = input().split()
values = map(int,input().split())

x = dict(zip(keys, values))
x = {key: value for key, value in x.items()
     if key != "delta" and value != 30}

print(x)
```
